```python
from sqlalchemy import create_engine,text
from sqlalchemy.engine import URL
import pandas as pd
```

```python
connection_url =  URL.create(
    drivername="postgresql+psycopg2",
    username="postgres",
    password=1234,
    host="localhost",
    port=5432,
    database="supply_chain_warehouse"
)
engine = create_engine(connection_url)
print("Database engine created successfully!")
```

```python
files = [
    "fact_shipment",
    "fact_production",
    "fact_procurement",
    "fact_inventory",
    "fact_sales",
    "dim_supplier",
    "dim_product",
    "dim_facility",
    "dim_date",
    "dim_customer"
]
```

```python
for file in files:
    df = pd.read_csv(f"../data/processed/{file}.csv")

    df.to_sql(
        file,
        engine,
        if_exists="replace",
        index=False
    )

    print(f"{file} loaded successfully!")

print("All files loaded successfully!")
```
